### Задача 4. Заказы в NumPy
Соберите из `rows` массивы `price`, `qty`, `product` (цикл для сбора списков можно, как на уроке). Посчитайте `revenue = price * qty`.
1. Сколько всего заказов?
2. Самый **дешёвый** заказ: сумма и товар (`argmin`).
3. Самый дорогой заказ: сумма. Во сколько раз он дороже самого дешёвого?

**Пример проверки:** `500` заказов; самый дешёвый — `7 500 ₸`, `Мышь Logitech M185`; самый дорогой — `1 098 000 ₸`.

````python
def fmt(x):
    """7500 -> '7 500'"""
    return f"{round(float(x)):,}".replace(",", " ")

price_list, qty_list, product_list = [], [], []
for r in rows:                      # цикл только для сбора списков
    price_list.append(int(r["price"]))
    qty_list.append(int(r["qty"]))
    product_list.append(r["product"])

price = np.array(price_list)
qty = np.array(qty_list)
product = np.array(product_list)
revenue = price * qty

n_orders = len(revenue)
i_min = revenue.argmin()
i_max = revenue.argmax()
ratio = revenue[i_max] / revenue[i_min]

print("Заказов:", n_orders)
print("Самый дешёвый:", fmt(revenue[i_min]), "₸ —", product[i_min])
print("Самый дорогой:", fmt(revenue[i_max]), "₸ —", product[i_max])
print(f"Дороже в {ratio:.1f} раза")
````

### Задача 6. Сколько потеряли на отменах
Соберите массив `status`. Посчитайте:
- сколько заказов отменено;
- какая это доля от всех заказов;
- на какую сумму были отменённые заказы.

Подсказка: `status == "отменён"` — массив `True/False`; `.sum()` считает `True`, `.mean()` даёт долю.

**Пример проверки:** `45` заказов, `9%`, `8 224 000 ₸`.

````python
status = np.array([r["status"] for r in rows])

cancelled = status == "отменён"
n_cancel = cancelled.sum()
share_cancel = cancelled.mean()
lost = revenue[cancelled].sum()

print("Отменено:", n_cancel)
print(f"Доля: {share_cancel:.0%}")
print("Сумма отмен:", fmt(lost), "₸")
````

### Задача 7. Неделя курьера
Курьер развозит по 5 заказов в день. Возьмите первые **35** заказов и разложите в таблицу «дни × заказы».
1. Какую форму передать в `reshape`?
2. Посчитайте сумму за каждый день — список из 7 чисел (цикл по строкам таблицы можно).
3. В какой день сумма была самой большой?

**Пример проверки:** самый «дорогой» день — **7-й**, `1 227 000 ₸`.

````python
week = revenue[:35].reshape(7, 5)      # 7 дней × 5 заказов
print(week.shape)

day_sums = []
for day in week:                        # цикл по строкам таблицы — можно
    day_sums.append(int(day.sum()))
day_sums = np.array(day_sums)

best_day = day_sums.argmax() + 1        # +1: дни считаем с 1
print("Суммы по дням:", day_sums)
print("Самый дорогой день:", best_day, "—", fmt(day_sums.max()), "₸")
````
